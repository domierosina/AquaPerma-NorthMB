```python
import geopandas as gpd
import folium
AOI_PATH = "../data/aoi/2025_aoi.geojson"   
# -------------------------------------------------------
# LOAD AOI
# -------------------------------------------------------
aoi_gdf = gpd.read_file(AOI_PATH)
centroid = [
    aoi_gdf.geometry.centroid.y.mean(),
    aoi_gdf.geometry.centroid.x.mean()
]

# -------------------------------------------------------
# BASE MAP (Google Satellite)
# -------------------------------------------------------
m = folium.Map(
    location=centroid,
    zoom_start=7,
    tiles=None
)

# Add Google Satellite tile layer
folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=s&x={x}&y={y}&z={z}",
    attr="Google Satellite",
    name="Google Satellite",
).add_to(m)

# -------------------------------------------------------
# AOI OVERLAY
# -------------------------------------------------------
folium.GeoJson(
    aoi_gdf,
    name="AOI",
    style_function=lambda x: {
        "color": "red",
        "weight": 4,
        "fillOpacity": 0.0
    }
).add_to(m)



# -------------------------------------------------------
# LAYER CONTROL & DISPLAY
# -------------------------------------------------------
folium.LayerControl().add_to(m)
m


m.save("AOI_map.html")
print("Saved AOI_map.html — open in browser and take a screenshot.")
```
