```python
import openeo
import json

# List candidate openEO back‑ends to test
BACKENDS = [
    "https://openeo.vito.be",                # VITO / Terrascope
    "https://openeo.cloud",                 # Copernicus / CDSE (federated)
    "https://openeo-dev.eodc.eu/openeo/1.1.0"  # EODC (if accessible; update to actual URL if changed)
]

# For each backend, test a short Sentinel‑2 collection + metadata filters
def test_backend(backend_url, collection_id="SENTINEL2_L2A"):
    print("\n=== Testing backend:", backend_url, "===\n")
    try:
        conn = openeo.connect(backend_url)
        conn = conn.authenticate_oidc()  # or authenticate_basic if required
    except Exception as e:
        print("  🔴 Connection failed:", e)
        return

    # 1. List available collections
    try:
        coll_ids = conn.list_collection_ids()
    except Exception as e:
        print("  ⚠️ Failed to list collections:", e)
        return

    print("  Collections available:", coll_ids)
    if collection_id not in coll_ids:
        print(f"  ⚠️ Collection {collection_id} not found on this backend.")
        return

    # 2. Describe collection (metadata)
    try:
        meta = conn.describe_collection(collection_id)
    except Exception as e:
        print("  ⚠️ describe_collection failed:", e)
        return

    # Pretty‑print some of metadata
    print("  Collection metadata (top level):")
    keys = list(meta.keys())
    print("   Keys:", keys)

    # 3. If the backend provides queryables, fetch them
    try:
        q = conn.get(f"/collections/{collection_id}/queryables")
        qj = q.json()
        print("  Queryables / metadata filters supported:")
        for prop, spec in qj.get("properties", {}).items():
            print("   ", prop)
    except Exception as e:
        print("  ⚠️ Could not fetch queryables metadata:", e)

    # 4. (Optional) Try a very small test load to see if cloud‑cover filtering works
    try:
        bbox = {"west": -97.0, "south": 49.5, "east": -96.5, "north": 49.9, "crs": "EPSG:4326"}
        t0, t1 = "2022-06-01", "2022-06-15"
        cube = conn.load_collection(
            collection_id=collection_id,
            spatial_extent=bbox,
            temporal_extent=[t0, t1],
            max_cloud_cover=20  # try to filter low‑cloud scenes
        )
        # Instead of downloading, request metadata only (this may or may not succeed)
        coll_meta = cube.metadata
        print("  Cube metadata returned:", type(coll_meta), coll_meta)
    except Exception as e:
        print("  ⚠️ Test load/filter failed (safe for large extents or metadata‑less backend):", e)

if __name__ == "__main__":
    for backend in BACKENDS:
        test_backend(backend)
```
